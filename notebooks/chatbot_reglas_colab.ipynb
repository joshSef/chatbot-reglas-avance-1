## 2. Inspeccionar la clasificación
El resultado muestra la intención elegida y la puntuación explicable de sus reglas.

```python
from asistente import RuleBasedChatbot

bot = RuleBasedChatbot(seed=7)
preguntas = [
    "Hola",
    "¿Cuál es el objetivo?",
    "¿Cómo funciona la arquitectura?",
    "¿Para qué sirve NLTK?",
    "¿Cómo genero evidencia?",
    "¿Cuál es el clima?",
]
for pregunta in preguntas:
    coincidencia = bot.classify(pregunta)
    print(f"Usuario: {pregunta}")
    print(f"Intención: {coincidencia.intent} | puntuación: {coincidencia.score}")
    print(f"Asistente: {bot.respond(pregunta)}\n")
```
